```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
image_path = '../cropped_images/image_1.png'
image = cv2.imread(image_path)
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
plt.imshow(image)
plt.show()
```

```python
half_body_image = image[0:int(image.shape[0]/2), :]
plt.imshow(half_body_image)
plt.show()
```

# cluster the image into two clusters

```python
# reshaping the image into 2d array
image_2d = half_body_image.reshape(-1, 3)

# perform K-means clustering with 2 clusters
kmeans = KMeans(n_clusters=2, random_state=0).fit(image_2d)

# get the cluster labels
labels = kmeans.labels_

# reshape the labels into the original image shape
clustered_image = labels.reshape(half_body_image.shape[0], half_body_image.shape[1])

# Draw the image
plt.imshow(clustered_image)
plt.show()
```

```python
corner_clusters = [clustered_image[0,0], clustered_image[-1,0], clustered_image[-1,-1], clustered_image[0,-1]]
non_player_cluster = max(set(corner_clusters), key=corner_clusters.count)
print(non_player_cluster)
```

```python
player_cluster = 1 - non_player_cluster
print(player_cluster)
```
